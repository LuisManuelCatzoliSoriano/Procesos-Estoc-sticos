# Ejercicio

El Sr. y la Sra. H han planeado celebrar su 25º aniversario de bodas en Honeymooners' Paradise, un popular complejo turístico en la ciudad de Centerville. Contando hoy como el primer día, se supone que estarán allí el séptimo y el octavo día. Están pensando en comprar un seguro de viaje que les reembolsará el costo total del paquete vacacional de $\$2500$ si llueve ambos días, y no se les reembolsará nada en caso contrario. El seguro cuesta $\$200$. Supongamos que el clima en la ciudad de Centerville cambia según el modelo de CMTD de tres estados (soleado, nublado y lluvioso) visto en clase. Suponiendo que hoy hace sol en la ciudad de Centerville, ¿deberían el Sr. y la Sra. H comprar el seguro?

$$P= \begin{pmatrix}
0.5 & 0.3 & 0.2 \\
0.5 & 0.2 & 0.3 \\
0.4 & 0.5 & 0.1 \\
\end{pmatrix}$$

Solución:

Defino los estados:

Para n=1,..., 8

$$x_n=\left\{\begin{matrix}
1 \hspace{0.2cm} \mathbf{si} \hspace{0.2cm} \mathbf{el} \hspace{0.2cm} \mathbf{día} \hspace{0.2cm} n \hspace{0.2cm} \mathbf{es} \hspace{0.2cm} \mathbf{soleado} \\ 2 \hspace{0.2cm} \mathbf{si} \hspace{0.2cm} \mathbf{el} \hspace{0.2cm} \mathbf{día} \hspace{0.2cm} n \hspace{0.2cm} \mathbf{es} \hspace{0.2cm} \mathbf{nublado}
 \\ 3 \hspace{0.2cm} \mathbf{si} \hspace{0.2cm} \mathbf{el} \hspace{0.2cm} \mathbf{día} \hspace{0.2cm} n \hspace{0.2cm} \mathbf{es} \hspace{0.2cm} \mathbf{lluvioso}
\end{matrix}\right.$$

Se tiene $\alpha=(1,0,0)$ y se debe calcular $P(x_7=3,x_8=3|x_1=1)$

$$P(x_7=3,x_8=3|x_1=1)=P(x_7=3|x_1=1)P(x_8=3|x_7=3)$$

$$=[P^6]_{1,3}P_{3,3}=(0.21)(0.1)$$

$$=0.021$$

Reembolso:

$$R=\left\{\begin{matrix}
2500 \hspace{0.2cm} \mathbf{si} \hspace{0.2cm} x7=x_8=3 \\ 0 \hspace{0.2cm} \mathbf{en} \hspace{0.2cm} \mathbf{otro} \hspace{0.2cm} \mathbf{caso}
\end{matrix}\right.$$

Entonces:

$$E(R)=2500P(x_7=x_8=3|x_1=1)=52.52$$

Como $\$52.52<\$200$, no convine porque implica una pérdida.
