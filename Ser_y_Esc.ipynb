# Serpientes y escaleras

image.png

¿Cuál es el número promedio de tiradas necesarias para terminar el juego de serpientes y escaleras en el tablero de la imagen adjunta?

```python
import random as r
import collections
import numpy as np
```

```python
FINAL = 20

# Corregir que si llega a la casilla 20 o mas con el dado se termina

jumps = {
    3: 11,   # Escalera
    15: 19,  # Escalera
    17: 9,   # Serpiente
    13: 4    # Serpiente
}


def serpientes_escaleras(max_pasos=1000):

    posicion = 0
    tiradas = 0

    while tiradas < max_pasos:

        # Tirar dado
        dado = np.random.randint(1, 7)

        nueva_pos = posicion + dado

        # Regla: debe caer exactamente en 20
        if nueva_pos > FINAL:
            nueva_pos = posicion

        # Aplicar serpientes/escaleras
        if nueva_pos in jumps:
            nueva_pos = jumps[nueva_pos]

        posicion = nueva_pos
        tiradas += 1

        # Verificar si terminó
        if posicion == FINAL:
            return tiradas

    return -1
```

```python
n = 10000

resultados = []

for _ in range(n):

    t = serpientes_escaleras()

    if t != -1:
        resultados.append(t)


promedio = np.mean(resultados)

print(f"Número de simulaciones: {n}")
print(f"Promedio de tiradas para terminar: {promedio:.4f}")
print(f"Mínimo de tiradas observado: {min(resultados)}")
print(f"Máximo de tiradas observado: {max(resultados)}")
```
