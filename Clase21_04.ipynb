# ¿Es regular?

```python
import numpy as np
import sympy as sp
```

```python
A=np.matrix([[0, 0, 0, 0, 1],
            [0.25, 0, 0, 0.5, 0.25],
            [0, 0, 1, 0, 0],
            [0, 0, 0.5, 0.5, 0],
            [0.25, 0.5, 0, 0, 0.25]])
```

```python
np.power(A, 3)
```

Si al elevar la matriz a dos potencias sucesivas los ceros se mantienen en la misma posición, entonces la matriz no es regular.

# Encontrar la distribución límite de

```python
B=sp.Matrix([[1/3, 1/2, 1/6],[1/2, 1/2, 0],[1/4, 1/2, 1/4]])
```

Primero verificamos que la matriz es regular

```python
np.power(B, 2)
```

```python
x, y, z = sp.symbols('x y z')
v=sp.Matrix([[x],[y],[z]])
r=B*v
```

```python
ec1=r[0, 0]
ec2=r[1, 0]
ec3=r[2, 0]
```

```python
e1=sp.Eq(ec1, x)
e2=sp.Eq(ec2, y)
e3=sp.Eq(ec3, z)
e4=sp.Eq(x+y+z,1)
```

```python
sol=sp.solve((e1,e2, e4), (x,y,z))
sol
```
