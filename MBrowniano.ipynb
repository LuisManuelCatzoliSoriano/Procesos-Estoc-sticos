1. Programar el seudocódigo y graficar una trayectoria para $T=1$ y $L=100$

```python
import numpy as np
import sympy as sp
import random as r
import matplotlib.pyplot as plt
```

```python
def MBrowniano(T,L):
  st=T/L
  t=np.linspace(0,T,L+1)
  W=np.zeros(L+1)
  for i in range (L):
    a=np.random.normal(0,1)
    W[i+1/2]=W[i]+a*np.sqrt(st)
  return t, W
```

```python
t,W=MBrowniano(1,100)
```

```python
plt.plot(t,W)
plt.xlabel('t')
plt.ylabel('W(t)')
plt.grid()
plt.show()
```

2. Graficar en el mismo panel 100 trayectorias junto con $f(t)=\sqrt(t)$, $g(t)=-\sqrt(t)$

```python
for i in range(100):
  t,W=MBrowniano(1,100)
  plt.plot(t,W, alpha=0.3)
plt.plot(t,np.sqrt(t),color='red', label=('sqrt(t)'), linewidth=3)
plt.plot(t,-np.sqrt(t),color='red', label=('-sqrt(t)'), linewidth=3)
plt.xlabel('t')
plt.ylabel('W(t)')
plt.grid()
plt.legend()
plt.show()
```

```python
tM,WM=MBrownianoM(1,100)
```
